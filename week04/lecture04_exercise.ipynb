## Scatter & Bubble Charts: Gapminder

> **Push to:** `week04/lecture04_exercise.ipynb`

**Rules:**
1. Colour used **sparingly** — one categorical variable, no rainbow
2. If showing all continents, either use accessible palette OR grey all + highlight one
3. `size_max` set when using bubble size
4. Log scale for GDP per capita
5. Insight title

---


```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2000 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1 — Scatter: life expectancy change over time (Asia 2002 vs 2007)

# 1. Filter data
df_task1 = df[(df['continent'] == 'Asia') & (df['year'].isin([2002, 2007]))]

# 2. Build the Scatter Plot
# Use color for year (exactly 2 colours)
fig = px.scatter(
    df_task1,
    x="gdpPercap",
    y="lifeExp",
    color="year",
    color_discrete_map={2002: "#A6CEE3", 2007: "#1F78B4"}, # Subtle vs Bold blue
    labels={"gdpPercap": "GDP per Capita", "lifeExp": "Life Expectancy"},
    log_x=True, # Rule 4: Log scale for GDP
    template="simple_white"
)

# 3. Apply Professional Styling
fig.update_layout(
    title="<b>Asian Nations Saw Consistent Gains in Health and Wealth Between 2002 and 2007</b>",
    font=dict(family="Arial", size=14, color="black"),
    showlegend=True,
    legend_title_text='Year',
    margin=dict(l=50, r=50, t=80, b=50)
)

fig.update_traces(marker=dict(size=10, opacity=0.8, line=dict(width=0.5, color='white')))

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2 — Bubble chart: tell a story (Highlighting Americas)

# 1. Prepare 2007 data
df_2007 = df[df['year'] == 2007].copy()

# 2. Define Highlight Color Logic (Rule 2)
highlight_continent = 'Americas'
df_2007['color_group'] = df_2007['continent'].apply(
    lambda x: highlight_continent if x == highlight_continent else 'Other'
)

# 3. Build Bubble Chart
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="color_group",
    color_discrete_map={highlight_continent: "#E31A1C", "Other": "#DDDDDD"}, # Grey + Highlight
    hover_name="country",
    size_max=60, # Rule 3: size_max set
    log_x=True,  # Rule 4: Log scale
    template="simple_white"
)

# 4. Add Annotation explaining the story
fig.add_annotation(
    x=4.1, xref="x", # Adjust based on log scale position
    y=75, yref="y",
    text="Most American nations maintain<br>high life expectancy (>70 years)<br>despite varying GDP levels.",
    showarrow=True,
    arrowhead=1,
    ax=40,
    ay=-60,
    font=dict(family="Arial", size=12)
)

# 5. Final Polish
fig.update_layout(
    title="<b>The Americas Reach High Life Expectancy Regardless of National Wealth</b><br>GDP vs Life Expectancy, size represents population (2007)",
    font=dict(family="Arial", size=14, color="black"),
    showlegend=False, # Removed clutter
    xaxis=dict(title="GDP per capita (Log Scale)", showgrid=False),
    yaxis=dict(title="Life Expectancy", showgrid=False),
    margin=dict(l=50, r=50, t=80, b=50)
)

fig.show()
```
